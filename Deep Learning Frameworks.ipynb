## 1. What is TensorFlow 2.0, and how is it different from TensorFlow 1.x

**Answer:**
TensorFlow 2.0 is a major version of TensorFlow that embraces **eager execution by default**, integrates tightly with Keras as the high-level API (`tf.keras`), and simplifies the user experience with clearer APIs. Differences from 1.x:

* **Eager execution** by default (no need to build `tf.Graph`/`Session` manually).
* **Keras-first**: `tf.keras` is the recommended interface for model building and training.
* Improved `tf.function` to convert Python functions to graph for speed.
* Cleaner APIs and removal/deprecation of many low-level TF1.x constructs.

## 3. What is the primary function of the `tf.function` in TensorFlow 2.0

**Answer:**
`tf.function` converts a Python function that uses TensorFlow ops into a **graph (a tf.Graph)**. This provides performance benefits (faster execution) and deployment advantages while letting you write code in eager mode. Example:

```python
import tensorflow as tf

@tf.function
def f(x):
    return x * x + 1

print(f(tf.constant(3)))
```

`tf.function` traces and compiles the function into graph mode automatically.

## 4. What is the purpose of the `Model` class in TensorFlow 2.0

**Answer:**
`tf.keras.Model` is the base class for creating models in Keras. It:

* Encapsulates layers and forward pass logic (`call` method).
* Provides training / evaluation / inference helpers (`compile`, `fit`, `evaluate`, `predict`).
* Manages saving/loading of weights and the whole model.
* Enables custom models by subclassing `tf.keras.Model`.

Example skeleton:

```python
class MyModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.d1 = tf.keras.layers.Dense(64, activation='relu')
        self.out = tf.keras.layers.Dense(10, activation='softmax')

    def call(self, inputs, training=False):
        x = self.d1(inputs)
        return self.out(x)
```

## 10. What is the difference between TensorFlow and PyTorch

**Answer (concise):**

* **Execution model:** Historically TF used static graphs (1.x) and PyTorch used eager; TF2 moved to eager by default and graph via `tf.function`. PyTorch is primarily eager/dynamic and intuitive debugging.
* **API style:** PyTorch is pythonic and popular in research; TensorFlow (with Keras) is production-oriented and has robust deployment tools (TF Serving, TFLite, TF.js).
* **Ecosystem:** TF has more built-in deployment options, TensorBoard; PyTorch has strong research adoption and supports TorchScript for tracing/serving.
* **Interchangeability:** Both are mature; choice often depends on personal/team preference and deployment constraints.

## 12. What is the basic structure of a PyTorch neural network

**Answer:**
A PyTorch model subclasses `torch.nn.Module` and defines layers in `__init__` and the forward pass in `forward`. Example:

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(784, 128)
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, 10)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        return self.fc3(x)
```

## Quick Practical
* **Choose optimizer**: `Adam` for fast convergence, `SGD` with momentum for some production cases.
* **Loss**: use `binary_crossentropy` for binary tasks, `sparse_categorical_crossentropy` for integer-labeled multiclass, `categorical_crossentropy` for one-hot labels.
* **Device**: use `.to(device)` in PyTorch to move tensors/models to GPU.
* **Reproducibility**: set random seeds (`tf.random.set_seed`, `torch.manual_seed`).
* **Monitoring**: prefer TensorBoard for TF and use `torch.utils.tensorboard` to write logs from PyTorch.
